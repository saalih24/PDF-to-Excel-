```python
import pdfplumber
import pandas as pd

pdf_path = "INELIGIBLE_APPLICATIONS_2024.pdf"
all_tables = []

with pdfplumber.open("C:\\Users\\Admin\\Downloads\\INELIGIBLE_APPLICATIONS_2024.pdf") as pdf:
    for page in pdf.pages:
        table = page.extract_table()
        if table:
            df = pd.DataFrame(table[1:], columns=table[0])
            all_tables.append(df)

# Combine all tables into one Excel sheet
final_df = pd.concat(all_tables, ignore_index=True)
final_df.to_excel("C:/Users/Admin/Downloads/output.xlsx", index=False)



```
